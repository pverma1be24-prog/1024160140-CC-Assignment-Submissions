### (d) Find the most frequent value and their indice(s) in the following arrays:
i. x = np.array([1,2,3,4,5,1,2,1,1,1])
ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def get_frequent_and_indices(arr):
    vals, counts = np.unique(arr, return_counts=True)
    max_count = np.max(counts)
    modes = vals[counts == max_count]
    res = {}
    for val in modes:
        indices = np.where(arr == val)[0]
        res[val] = indices
    return res

print("Array x analysis:")
for val, idxs in get_frequent_and_indices(x).items():
    print(f"  Value {val} is most frequent. Indices: {idxs}")

print("\nArray y analysis:")
for val, idxs in get_frequent_and_indices(y).items():
    print(f"  Value {val} is most frequent. Indices: {idxs}")
```

### (f) For the matrix: n_array = np.array([[55, 25, 15],[30, 44, 2],[11, 45, 77]]), find:
i. Sum of diagonal elements
ii. Eigen values of matrix
iii. Eigen vectors of matrix
iv. Inverse of matrix
v. Determinant of matrix

```python
n_array = np.array([[55, 25, 15], [30, 44, 2], [11, 45, 77]])
diag_sum = np.trace(n_array)
eigenvals, eigenvecs = np.linalg.eig(n_array)
inv_mat = np.linalg.inv(n_array)
det_mat = np.linalg.det(n_array)
print("Diagonal Sum:", diag_sum)
print("Eigen Values:", eigenvals)
print("Eigen Vectors:\n", eigenvecs)
print("Inverse:\n", inv_mat)
print("Determinant:", det_mat)
```

### (g) Multiply the following matrices and also find covariance between matrices using NumPy:
i. p = [[1, 2], [2, 3]], q = [[4, 5], [6, 7]]
ii. p = [[1, 2], [2, 3], [4, 5]], q = [[4, 5, 1], [6, 7, 2]]

```python
p1 = np.array([[1, 2], [2, 3]])
q1 = np.array([[4, 5], [6, 7]])
product1 = p1 @ q1
covariance1 = np.cov(p1.flatten(), q1.flatten())
print("i. Matrix Multiplication Result:\n", product1)
print("i. Covariance Between Matrices (Flattened):\n", covariance1)

p2 = np.array([[1, 2], [2, 3], [4, 5]])
q2 = np.array([[4, 5, 1], [6, 7, 2]])
product2 = p2 @ q2
covariance2 = np.cov(p2.flatten(), q2.flatten())
print("\nii. Matrix Multiplication Result:\n", product2)
print("ii. Covariance Between Matrices (Flattened):\n", covariance2)
```

### (h) For the matrices: x = np.array([[2, 3, 4], [3, 2, 9]]); y = np.array([[1, 5, 0], [5, 10, 3]]), find inner, outer and cartesian product?

```python
x = np.array([[2, 3, 4], [3, 2, 9]])
y = np.array([[1, 5, 0], [5, 10, 3]])
inner_prod = np.inner(x, y)
outer_prod = np.outer(x, y)
x_flat = x.flatten()
y_flat = y.flatten()
cartesian_prod = np.transpose([np.repeat(x_flat, len(y_flat)), np.tile(y_flat, len(x_flat))])
print("Inner Product:\n", inner_prod)
print("\nOuter Product:\n", outer_prod)
print("\nCartesian Product (First 10 entries):\n", cartesian_prod[:10])
```

## Q2: Based on NumPy Mathematics and Statistics

### (a) For the array: array = np.array([[1, -2, 3],[-4, 5, -6]])
i. Find element-wise absolute value
ii. Find the 25th, 50th, and 75th percentile of flattened array, for each column, for each row.
iii. Mean, Median and Standard Deviation of flattened array, of each column, and each row

```python
array = np.array([[1, -2, 3], [-4, 5, -6]])

abs_val = np.abs(array)
print("i. Element-wise Absolute Value:\n", abs_val)

percentiles_flat = np.percentile(array, [25, 50, 75])
percentiles_col = np.percentile(array, [25, 50, 75], axis=0)
percentiles_row = np.percentile(array, [25, 50, 75], axis=1)
print("\nii. Percentiles (25th, 50th, 75th):")
print("  Flattened:", percentiles_flat)
print("  Column-wise:\n", percentiles_col)
print("  Row-wise:\n", percentiles_row)

mean_flat, mean_col, mean_row = np.mean(array), np.mean(array, axis=0), np.mean(array, axis=1)
median_flat, median_col, median_row = np.median(array), np.median(array, axis=0), np.median(array, axis=1)
std_flat, std_col, std_row = np.std(array), np.std(array, axis=0), np.std(array, axis=1)
print("\niii. Statistics (Mean, Median, Standard Deviation):")
print(f"  Flattened: Mean = {mean_flat}, Median = {median_flat}, Std = {std_flat}")
print(f"  Column-wise: Mean = {mean_col}, Median = {median_col}, Std = {std_col}")
print(f"  Row-wise: Mean = {mean_row}, Median = {median_row}, Std = {std_row}")
```

### (a) Write a function named img_to_array(path) that reads an image from a specified path and save it as text file on local machine? (Note: use separate cases for RGB and Grey Scale images)
### (b) Load the saved file into jupyter notebook?

```python
from PIL import Image

def img_to_array(path, txt_path="image_data.txt"):
    img = Image.open(path)
    img_arr = np.array(img)
    if img_arr.ndim == 2:
        np.savetxt(txt_path, img_arr, fmt='%d')
        print(f"Saved Grayscale image of shape {img_arr.shape} to {txt_path}")
    elif img_arr.ndim == 3:
        h, w, c = img_arr.shape
        flattened = img_arr.reshape(-1, c)
        header = f"{h},{w},{c}"
        np.savetxt(txt_path, flattened, fmt='%d', header=header)
        print(f"Saved RGB image of shape {img_arr.shape} to {txt_path}")
    else:
        raise ValueError("Unsupported image dimensions")

dummy_rgb = np.random.randint(0, 256, (5, 5, 3), dtype=np.uint8)
Image.fromarray(dummy_rgb).save("dummy_rgb.png")

dummy_gray = np.random.randint(0, 256, (5, 5), dtype=np.uint8)
Image.fromarray(dummy_gray).save("dummy_gray.png")

img_to_array("dummy_rgb.png", "dummy_rgb.txt")
img_to_array("dummy_gray.png", "dummy_gray.txt")

loaded_gray = np.loadtxt("dummy_gray.txt", dtype=int)
print("\nLoaded Grayscale Image:\n", loaded_gray)

loaded_rgb_raw = np.loadtxt("dummy_rgb.txt", dtype=int)
loaded_rgb = loaded_rgb_raw.reshape(5, 5, 3)
print("\nLoaded RGB Image (first channel):\n", loaded_rgb[:, :, 0])
```
