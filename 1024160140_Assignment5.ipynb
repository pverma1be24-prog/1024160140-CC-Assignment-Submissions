### Question 1
Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:
a) Addition of 2 in all the elements
b) Multiply 3 with all the elements
c) Divide every element by 2

```python
import numpy as np

# Create a NumPy 1D-array with 5 elements
arr1 = np.array([10, 20, 30, 40, 50])
print("Original 1D-Array:", arr1)

# a) Addition of 2 in all the elements
arr_add = arr1 + 2
print("a) Addition of 2 in all elements:", arr_add)

# b) Multiply 3 with all the elements
arr_mul = arr1 * 3
print("b) Multiply 3 with all elements:", arr_mul)

# c) Divide every element by 2
arr_div = arr1 / 2
print("c) Divide every element by 2:", arr_div)
```

### Question 2
Questions on Basic NumPy Array:
a) Reverse the NumPy array: `arr = np.array([1, 2, 3, 6, 4, 5])`
b) Find the most frequent value and their indice(s) in the following arrays:
   i. `x = np.array([1,2,3,4,5,1,2,1,1,1])`
   ii. `y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])`

```python
# a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("Original Array arr:", arr)
print("Reversed Array:", reversed_arr)

print("\n" + "="*50 + "\n")

# b) Find the most frequent value and their indice(s)
# i. x = np.array([1,2,3,4,5,1,2,1,1,1])
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values_x, counts_x = np.unique(x, return_counts=True)
max_count_x = np.max(counts_x)
most_frequent_x = values_x[counts_x == max_count_x]

print("Array x:", x)
for val in most_frequent_x:
    indices_x = np.where(x == val)[0]
    print(f"Most frequent value in x: {val} (appears {max_count_x} times) at index/indices: {indices_x}")

print("\n" + "="*50 + "\n")

# ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values_y, counts_y = np.unique(y, return_counts=True)
max_count_y = np.max(counts_y)
most_frequent_y = values_y[counts_y == max_count_y]

print("Array y:", y)
for val in most_frequent_y:
    indices_y = np.where(y == val)[0]
    print(f"Most frequent value in y: {val} (appears {max_count_y} times) at index/indices: {indices_y}")
```

### Question 4
Write program to create an 1-D NumPy array named `prashant_narayan_verma` with evenly spaced 25 numbers from 10 to 100 using `linspace()`. Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using `reshape()` attribute. Can we do the same with `T` attribute?

```python
# Create 1-D array using linspace
prashant_narayan_verma = np.linspace(10, 100, 25)
print("1-D Array (prashant_narayan_verma):\n", prashant_narayan_verma)

# Print attributes
print("\nDimensions of array:", prashant_narayan_verma.ndim)
print("Shape of array:", prashant_narayan_verma.shape)
print("Total number of elements:", prashant_narayan_verma.size)
print("Data type of each element:", prashant_narayan_verma.dtype)
print("Total number of bytes consumed:", prashant_narayan_verma.nbytes)

# Find transpose using reshape() attribute
# Reshaping to (25, 1) converts the 1D array into a 2D column vector, representing the transpose.
transposed_reshape = prashant_narayan_verma.reshape(-1, 1)
print("\nTranspose using reshape(-1, 1) (Column Vector Shape):\n", transposed_reshape)
print("Shape after reshape:", transposed_reshape.shape)

# Explanation for T attribute:
# Can we do the same with T attribute?
# Direct application of .T on a 1-D array does not change its shape because a 1-D array is a single vector of shape (25,) 
# and has no second axis to swap. Therefore, prashant_narayan_verma.T is identical to prashant_narayan_verma.
#
# To transpose to a 2D column vector using .T, we must first add a new dimension to make it a 2D row vector (shape (1, 25)), 
# and then transpose it.
transposed_T = prashant_narayan_verma[np.newaxis, :].T
print("\nDirect transpose (prashant_narayan_verma.T) shape:", prashant_narayan_verma.T.shape)
print("Using np.newaxis and .T to transpose:\n", transposed_T)
print("Shape:", transposed_T.shape)
```

### Question 5
Create a 2-D Array of three rows and four columns, named `ucs420_prashant` with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as `reshaped_ucs420_prashant`. Resize the array to two rows and three columns and name it as `resized_ucs420_prashant`.

```python
# Create the 2-D array of 3 rows and 4 columns
values = [10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]
ucs420_prashant = np.array(values).reshape(3, 4)
print("Original 2-D Array (ucs420_prashant):\n", ucs420_prashant)

# Compute statistics
print("\n--- Statistics ---")
print("Mean:", np.mean(ucs420_prashant))
print("Median:", np.median(ucs420_prashant))
print("Max:", np.max(ucs420_prashant))
print("Min:", np.min(ucs420_prashant))
print("Unique elements:", np.unique(ucs420_prashant))

# Reshape the array to 4 rows and 3 columns (named reshaped_ucs420_prashant)
reshaped_ucs420_prashant = ucs420_prashant.reshape(4, 3)
print("\n--- Reshaped Array (4x3) ---")
print(reshaped_ucs420_prashant)

# Resize the array to 2 rows and 3 columns (named resized_ucs420_prashant)
# Note: np.resize creates a new array of the specified shape. Since 2x3 needs 6 elements, 
# it will take the first 6 elements of the original array.
resized_ucs420_prashant = np.resize(ucs420_prashant, (2, 3))
print("\n--- Resized Array (2x3) ---")
print(resized_ucs420_prashant)
```
