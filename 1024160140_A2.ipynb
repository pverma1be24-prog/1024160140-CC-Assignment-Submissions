## List Operations

```python
import random

roll_no = "1024160140"
L = [int(digit) * 10 for digit in roll_no]
print("i. Original List L:", L)
```

```python
L.append(99)
print("ii. L after append(99):", L)
```

```python
L.insert(3, 55)
print("ii. L after insert(3, 55):", L)
```

```python
L.remove(10)
print("iii. L after remove(10):", L)

popped_val = L.pop()
print("iii. L after pop():", L)
```

```python
L.sort()
print("iv. L sorted ascending:", L)

L.sort(reverse=True)
print("iv. L sorted descending:", L)
```

```python
print("v. First three elements:", L[:3])
print("v. Last three elements:", L[-3:])
```

```python
avg_L = sum(L) / len(L)
L_above_avg = [x for x in L if x > avg_L]
print("vi. Elements greater than average of L:", L_above_avg)
```

## Tuple Operations

```python
scores = tuple(L[:8])
print("\nScores Tuple:", scores)
```

```python
highest = max(scores)
highest_idx = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print(f"i. Highest score: {highest} at index {highest_idx}")
print(f"i. Lowest score: {lowest} appearing {lowest_count} times")
```

```python
reversed_scores = list(reversed(scores))
print("ii. Reversed list:", reversed_scores)
```

```python
user_input = input("Enter a score to search: ")
try:
    search_score = int(user_input)
    if search_score in scores:
        print(f"iii. Score {search_score} first occurs at index {scores.index(search_score)}")
    else:
        print(f"iii. Score {search_score} is not present in the tuple")
except ValueError:
    print("iii. Invalid input, not an integer")
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print("iv. Error raised:", e)
```

```python
first, second, *remaining = scores
print(f"v. First: {first}, Second: {second}, Remaining: {remaining}")
```

## Random Number Generation and Analysis

```python
random.seed(1024160140)
random_list = [random.randint(100, 900) for _ in range(100)]
print("\ni. Generated 100 random numbers (first 10 elements):", random_list[:10])
```

```python
odd_numbers = [x for x in random_list if x % 2 != 0]
print(f"ii. Odd numbers count: {len(odd_numbers)}")
print("ii. Odd numbers:", odd_numbers)
```

```python
even_numbers = [x for x in random_list if x % 2 == 0]
print(f"iii. Even numbers count: {len(even_numbers)}")
print("iii. Even numbers:", even_numbers)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_list if is_prime(x)]
print(f"iv. Prime numbers count: {len(prime_numbers)}")
print("iv. Prime numbers:", prime_numbers)
```

```python
most_frequent = max(set(random_list), key=random_list.count)
freq = random_list.count(most_frequent)
print(f"v. Most frequent number: {most_frequent} (occurs {freq} times)")
```

## Set Operations

```python
digits = [1, 0, 2, 4, 1, 6, 0, 1]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}
print("\nSet A:", A)
print("Set B:", B)
```

```python
union_AB = A.union(B)
print("vi. Union of A and B:", union_AB)
```

```python
intersection_AB = A.intersection(B)
print("vii. Intersection of A and B:", intersection_AB)
```

```python
diff_AB = A.difference(B)
diff_BA = B.difference(A)
print("viii. A - B:", diff_AB)
print("viii. B - A:", diff_BA)
```

```python
sym_diff = A.symmetric_difference(B)
print("ix. Symmetric Difference:", sym_diff)
```

```python
print("x. Is A subset of B?:", A.issubset(B))
print("x. Is B superset of A?:", B.issuperset(A))
```

```python
val_x_input = input("Enter a value to discard from A: ")
try:
    val_x = int(val_x_input)
    A.discard(val_x)
    print(f"xi. Set A after discarding {val_x}:", A)
except ValueError:
    print("xi. Invalid input, not an integer")
```
