```python
# Creating the sample dataset for Question 6
data_q6 = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

df_employees = pd.DataFrame(data_q6)

# Saving to employees.csv
df_employees.to_csv('employees.csv', index=False)
df_employees
```

#### e) Calculate the following statistics from the dataset:
1. The average salary of employees.
2. The total bonus paid to all employees.
3. The youngest employee's age.
4. The highest performance rating.

```python
# i. Average salary
avg_salary = df_employees['Salary'].mean()
# ii. Total bonus
total_bonus = df_employees['Bonus'].sum()
# iii. Youngest employee's age
youngest_age = df_employees['Age'].min()
# iv. Highest performance rating
highest_rating = df_employees['Rating'].max()

print(f"i. Average Salary of Employees: ${avg_salary:,.2f}")
print(f"ii. Total Bonus Paid to all Employees: ${total_bonus:,.2f}")
print(f"iii. Youngest Employee's Age: {youngest_age} years")
print(f"iv. Highest Performance Rating: {highest_rating}")
```

#### f) Sort the DataFrame by the Salary column in descending order.

```python
# Sorting by Salary in descending order
df_sorted = df_employees.sort_values(by='Salary', ascending=False)
df_sorted
```

#### g) Add a new column that categorizes employees based on their performance rating:
- **Excellent** for ratings >= 4.5
- **Good** for ratings >= 4.0 but < 4.5
- **Average** for ratings < 4.0

```python
import numpy as np

# Defining the conditions and labels
conditions = [
    df_employees['Rating'] >= 4.5,
    (df_employees['Rating'] >= 4.0) & (df_employees['Rating'] < 4.5),
    df_employees['Rating'] < 4.0
]
categories = ['Excellent', 'Good', 'Average']

# Adding new Performance_Category column
df_employees['Performance_Category'] = np.select(conditions, categories, default='Average')
df_employees[['Name', 'Rating', 'Performance_Category']]
```

#### h) Identify missing values in the DataFrame.

```python
# Identifying missing values per column
missing_summary = df_employees.isnull().sum()
print("Missing values in each column:")
print(missing_summary)
```

#### i) Rename the Employee_ID column to ID.

```python
# Renaming 'Employee_ID' to 'ID'
df_employees.rename(columns={'Employee_ID': 'ID'}, inplace=True)
df_employees
```

#### j) Find all employees who:
1. Have more than 5 years of experience.
2. Belong to the IT department.

```python
print("i. Employees with > 5 years of experience:")
experience_filter = df_employees[df_employees['Years_of_Experience'] > 5]
display(experience_filter)

print("\nii. Employees belonging to the IT department:")
it_filter = df_employees[df_employees['Department'] == 'IT']
display(it_filter)

print("\nCombined filter (IT Department and > 5 years of experience):")
combined_filter = df_employees[(df_employees['Department'] == 'IT') & (df_employees['Years_of_Experience'] > 5)]
display(combined_filter)
```

#### k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.

```python
# Deduct 10% of the Salary and store in new column 'Tax'
df_employees['Tax'] = df_employees['Salary'] * 0.10
df_employees
```
